```python
from random import random
from math import sqrt
def g(x,y,z):
  return sqrt(3-x**2-y**2-z**2)
N=1000
suma=0
for i in range(N):
  x=random()
  y=random()
  z=random()
  suma=suma+g(x,y,z)
print(suma/N)
```
