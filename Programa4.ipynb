```python
from math import log
import numpy as np

def f(p):
  """Generates a single sample from a geometric distribution with probability p."""
  u = np.random.uniform(0, 1)
  return 1 + int(log(u) / log(1 - p))

p = .7
N = 1000
suma = 0
for i in range(N):
  g = f(p)
  suma = suma + g
print(suma / N, 1 / p)
```
